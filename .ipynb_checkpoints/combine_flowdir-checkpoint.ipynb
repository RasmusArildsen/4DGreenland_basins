```python
# -----------------------------
# Inputs
# -----------------------------
Surface_dem = r"C:\Users\B464518\drainage\systematisk\dems\arcticDEM_500m_ice_sheet.tif"
Bed_dem     = r"C:\Users\B464518\drainage\systematisk\hydro_pot\arctic_s2b1.tif"
poly_2000m  = r"C:\Users\B464518\drainage\systematisk\2000m_polygon.shp"

# Import polygon and DEMs
gs.run_command('v.import', input=poly_2000m.replace("\\","/"), output='polygon_area', overwrite=True)
gs.run_command('r.import', input=Surface_dem, output='surface_dem', overwrite=True)
gs.run_command('r.import', input=Bed_dem, output='bed_dem', overwrite=True)

#smooth surface_dem
gs.run_command('r.neighbors',
                input='surface_dem',
                output='surface_dem_smooth',
                method='average',
                size=3,
                overwrite=True)
```

```python
gs.run_command("v.to.rast", input="polygon_area", output="inside_mask",
               use="val", value=1, overwrite=True)

# Sørg for at masken er =1 og alt andet er =0, ikke null.

gs.run_command("r.mapcalc", expression="inside_mask = if(isnull(inside_mask), 0, inside_mask)",
               overwrite=True)

# outside mask = 1 outside polygon
gs.run_command("r.mapcalc", expression="outside_mask = if(inside_mask == 0, 1, 0)",
               overwrite=True)

# Surface DEM only inside polygon
gs.run_command("r.mapcalc", expression="surface_masked = if(inside_mask == 1, surface_dem_smooth, null())",
               overwrite=True)

# Bed DEM only outside polygon
gs.run_command("r.mapcalc", expression="bed_masked = if(outside_mask == 1, bed_dem, null())",
               overwrite=True)

## Fill DEMS separately 
gs.run_command("r.fill.dir", input="surface_masked", output="dem_filledA",
               direction="flow_dirA", overwrite=True)
gs.run_command("r.fill.dir", input="bed_masked", output="dem_filledB",
               direction="flow_dirB", overwrite=True)

#combine the two filled-dems##
gs.run_command("r.mapcalc", expression="hybrid_dem = if(inside_mask == 1, dem_filledA, dem_filledB)",
               overwrite=True)

#----------------------------------------####### SKAL VI FILL SINKS IGEN HER? ELLER IKKE? #######---------------------------------------#
gs.run_command("r.watershed", elevation="hybrid_dem", accumulation="accum", overwrite=True)

gs.run_command("r.stream.extract", elevation="hybrid_dem", direction="flow_dir_hybrid",
               accumulation="accum", threshold=STREAM_THRESHOLD,
               d8cut=0, mexp=0.0, stream_raster="streams", overwrite=True)
gs.run_command('r.stream.basins',
               direction='flow_dir_hybrid',
               basins='basins',
               #points ='outlets',
               stream_rast='streams',
               flags = 'l',
               overwrite=True)
```

```python
#----------eksporter----------#
for name, fn in [
    ("accum", fr"{OUT}\basins\combined\accum_hybrid_FD.tif"),
    ("flow_dir", fr"{OUT}\basins\combined\flowdir_hybrid_FD.tif"),
    ("streams", fr"{OUT}\basins\combined\streams_hybrid_FD.tif"),
    ("basins", fr"{OUT}\basins\combined\basins_hybrid_FD.tif"),
]:
    gs.run_command("r.out.gdal", input=name, output=fn.replace("\\","/"), format="GTiff",
                   createopt="COMPRESS=LZW", overwrite=True)
```
