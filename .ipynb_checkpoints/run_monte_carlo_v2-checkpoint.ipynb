```python
# run_monte_carlo.py
from pathlib import Path

from mc_postprocess_2 import (
    build_mc_products,   # optional at end
    make_perturbed_dem,
    make_perturbed_bed,
)

import basin_core as bc
from basin_core import (
    setup_grass_env,
    start_grass_from_raster,
    prepare_dem_and_mask,
    ensure_grass_addon,
    run_hydro_for_dem,
    build_hybrid_dem_for_run_mask,   
    import_dem_native,
    QGIS_PREFIX_DEFAULT,
)


# --------------------------------------------------------------------
# USER SETTINGS
# --------------------------------------------------------------------
dem_res = 500
dem_type = "bed"  # 'surf' | 'bed' | 'hybrid'

if dem_res == 500:
    DEM = r"/Users/ralor/Rasmus/DTU/Cryo/DEM/arcticDEM_500m_ice_sheet.tif"
    VAR = r"/Users/ralor/Rasmus/DTU/Cryo/DEM/arcticdem_mosaic_500m_v4.1/arcticdem_mosaic_500m_v4.1_sigma_ice_masked.tif"
    fill_name = "dem_filled_500m"
    STREAM_THRESHOLD = 200
    CORR_PIX = 5

    if dem_type == "surf":
        OUT = r"/Volumes/Lenovo PS8/4DBasins/arcticdem500_mc"
    elif dem_type == "bed":
        OUT = r"/Volumes/Lenovo PS8/4DBasins/bed_machine_500_mc"
    else:
        OUT = r"/Volumes/Lenovo PS8/4DBasins/hybrid_500_mc2"

elif dem_res == 100:
    DEM = r"/Users/ralor/Rasmus/DTU/Cryo/DEM/arcticDEM_100m/arcticdem_mosaic_100m_v4.1_dem_clipped.tif"
    VAR = r"/Users/ralor/Rasmus/DTU/Cryo/DEM/arcticDEM_100m/arcticdem_mosaic_100m_v4.1_sigma_ice_masked.tif"
    fill_name = r"dem_filled_100m"
    STREAM_THRESHOLD = 5000
    CORR_PIX = 5

    if dem_type == "surf":
        OUT = r"/Volumes/Lenovo PS8/4DBasins/arcticdem100_mc"
    elif dem_type == "bed":
        OUT = r"/Volumes/Lenovo PS8/4DBasins/bed_machine_100_mc"
    else:
        OUT = r"/Volumes/Lenovo PS8/4DBasins/hybrid_100_mc"

else:
    raise ValueError("dem_res must be 100 or 500")

BASIN_PATTERN = "basins_hydro_mc_*.tif"
ICE_MASK = r"/Users/ralor/Rasmus/DTU/Jupyter_Notebook/4DGreenland_basins/data/ice_mask/02-PROMICE-2022-IceMask-polygon.gpkg"

N_MC = 500
START_I = 1

FILL_HOLES = False
FILL_METHOD = "bilinear"

# hydrology smoothing in run_hydro_for_dem (NOT used for hybrid; hybrid does its own smoothing)
SMOOTH = False
SMOOTH_SIZE = 3

QGIS_PREFIX = QGIS_PREFIX_DEFAULT

# Hybrid inputs (boundary between surface-use and bed-use)
POLY_2000M = r"/Volumes/Lenovo PS8/4DBasins/til_rasmus/til_rasmus_ensemble/2000m_isoline.shp"
if dem_res == 500:
    BED_FILE   = r"/Users/ralor/Rasmus/DTU/Cryo/DEM/bedmachine_v6_500m.tif"
    #r"/Volumes/Lenovo PS8/4DBasins/til_rasmus/til_rasmus_ensemble/bedmachine_500m.tif"
    BED_ERR    = r"/Users/ralor/Rasmus/DTU/Cryo/DEM/bedmachine_v6_error_500m.tif"
    #r"/Volumes/Lenovo PS8/4DBasins/til_rasmus/til_rasmus_ensemble/err_bed_500m.tif"
    BED_ERR_IS_VARIANCE = False  # True if bed_err is σ²

if dem_res == 100:
    BED_FILE   = r"/Volumes/Lenovo PS8/4DBasins/til_rasmus/til_rasmus_ensemble/bedmachine_100m.tif"
    BED_ERR    = r"/Volumes/Lenovo PS8/4DBasins/til_rasmus/til_rasmus_ensemble/bedmachine_errbed_100m.tif"
    BED_ERR_IS_VARIANCE = False  # True if bed_err is σ²

# Optional exports per run
EXPORT_DEBUG = True
EXPORT_RUNS  = {11,30,31}     # set None to export all, or e.g. {11,12,50}
EXPORT_EVERY = 25       # only used if EXPORT_RUNS is None

Path(OUT).mkdir(parents=True, exist_ok=True)

# --------------------------------------------------------------------
# INIT GRASS
# --------------------------------------------------------------------
setup_grass_env(QGIS_PREFIX)
gs = bc.gs  # GRASS module initialised by setup_grass_env

MAPSET = f"MC_WORK_{dem_res}m"
GISDBASE, LOCATION, MAPSET = start_grass_from_raster(DEM, location="dem_loc", mapset=MAPSET)

ensure_grass_addon("r.stream.extract")
ensure_grass_addon("r.stream.basins")

# Create & switch to a work mapset so PERMANENT stays clean
gs.run_command("g.mapset", flags="c", mapset=MAPSET)

# --------------------------------------------------------------------
# HELPERS FOR CLEANUP + EXPORTS
# --------------------------------------------------------------------
def list_maps(map_type: str) -> set[str]:
    txt = gs.read_command("g.list", type=map_type, separator="newline")
    return {ln.strip() for ln in txt.splitlines() if ln.strip()}

def remove_maps(map_type: str, names: list[str]):
    if not names:
        return
    gs.run_command("g.remove", flags="f", type=map_type, name=",".join(names))

def raster_exists(name: str) -> bool:
    txt = gs.read_command("g.findfile", element="cell", file=name)
    return ("name=" in txt) and not ("name=\n" in txt or "name=\r\n" in txt)

def export_raster(name: str, out_path: Path, *, gdal_type: str | None = None, nodata=None):
    out_path.parent.mkdir(parents=True, exist_ok=True)

    # Remove color tables to avoid SetColorTable warnings for Int32 exports
    try:
        gs.run_command("r.colors", flags="r", map=name)
    except Exception:
        pass

    kwargs = dict(
        input=name,
        output=out_path.as_posix(),
        format="GTiff",
        createopt="COMPRESS=LZW,TILED=YES,BIGTIFF=YES",
        overwrite=True,
        flags="f",  # <-- override precision-loss abort
    )
    if gdal_type is not None:
        kwargs["type"] = gdal_type
    if nodata is not None:
        kwargs["nodata"] = nodata

    gs.run_command("r.out.gdal", **kwargs)

def export_maps_for_run(i: int, out_dir: str):
    out_dir = Path(out_dir)
    tag_h = f"_mc_{i:03d}"   # hydrology names inside run_hydro_pipeline
    tag   = f"_{i:03d}"      # hybrid names inside build_hybrid_dem_for_run

    exports = [
        # hydrology products
        (f"accum{tag_h}",    out_dir / f"accum_hybrid{tag_h}.tif",    None),
        (f"flow_dir{tag_h}", out_dir / f"flowdir_hybrid{tag_h}.tif",  "Int32"),
        (f"streams{tag_h}",  out_dir / f"streams_hybrid{tag_h}.tif",  "Int16"),
        (f"basins{tag_h}",   out_dir / f"basins_hybrid{tag_h}.tif",   "Int32"),

        # hybrid intermediates
        (f"hybrid_dem{tag}",     out_dir / f"hybrid_dem{tag}.tif",     "Float64"),
        (f"bed_masked{tag}",     out_dir / f"bed_masked{tag}.tif",     "Float64"),
        (f"surface_masked{tag}", out_dir / f"surface_masked{tag}.tif", "Float64"),
        (f"dem_filledA{tag}",    out_dir / f"dem_filledA{tag}.tif",    "Float64"),
        (f"dem_filledB{tag}",    out_dir / f"dem_filledB{tag}.tif",    "Float64"),
    ]

    for name, fn, gtype in exports:
        if not raster_exists(name):
            print(f"⚠️ Missing raster (skip): {name}")
            continue
        export_raster(name, fn, gdal_type=gtype, nodata=-9999)
        print(f"📤 Exported {name} → {fn}")

# --------------------------------------------------------------------
# DEM + MASK
# --------------------------------------------------------------------
dem_map, ice_mask_rast = prepare_dem_and_mask(
    dem_path=DEM,
    ice_mask_path=ICE_MASK,
    fill_holes=FILL_HOLES,
    fill_method=FILL_METHOD,
    filled_name=fill_name,
)

# variance for surface perturbation
import_dem_native(VAR, out_name="dem_var")

# --------------------------------------------------------------------
# IMPORT HYBRID ANCILLARIES ONCE (IN CURRENT MAPSET)
# --------------------------------------------------------------------
SURF_2000 = r"/Users/ralor/Rasmus/DTU/Cryo/DEM/arcticDEM_500m_ice_sheet_2000m.tif"

# Bed + bed error (for perturbations)
import_dem_native(BED_FILE, out_name="bed_raster")
import_dem_native(BED_ERR,  out_name="bed_err")

# 2000m mask raster (values >=2000, NULL below)
import_dem_native(SURF_2000, out_name="surf_mask_2000")   # <-- THIS NAME is used later


POLY_VEC = "switch_2000m"
gs.run_command("v.import", input=POLY_2000M.replace("\\", "/"), output=POLY_VEC, overwrite=True)

# --------------------------------------------------------------------
# KEEP SETS (for cleanup)
# --------------------------------------------------------------------

KEEP_RASTERS = {
    dem_map, "dem_var", ice_mask_rast, fill_name, "MASK",
    "bed_raster", "bed_err", "surf_mask_2000"
}

# prepare_dem_and_mask creates ice_mask_vec
KEEP_VECTORS = {"ice_mask_vec"}
```

```python
# --------------------------------------------------------------------
# MONTE CARLO LOOP
# --------------------------------------------------------------------
for i in range(START_I, N_MC + 1):
    print(f"\n==================== Monte Carlo Run {i:03d}/{N_MC} ====================")

    ras_before = list_maps("raster")
    vec_before = list_maps("vector")

    # 1) Perturbed surface DEM
    dem_mc = make_perturbed_dem(
        run_idx=i,
        base_dem=dem_map,
        var_map="dem_var",
        corr_pix=CORR_PIX,
    )

    # 2) Perturbed bed DEM
    bed_mc, pert_bed = make_perturbed_bed(
        run_idx=i,
        base_bed="bed_raster",
        err_map="bed_err",
        corr_pix=CORR_PIX,
        err_is_variance=BED_ERR_IS_VARIANCE,
    )

    # 3) Decide which DEM to run hydrology on
    if dem_type == "surf":
        dem_for_hydro = dem_mc
    elif dem_type == "bed":
        dem_for_hydro = bed_mc
    else:
        tag = f"_{i:03d}"
        dem_for_hydro = build_hybrid_dem_for_run_mask(
            surface_dem=dem_mc,
            bed_raster=bed_mc,
            inside_mask_raster="surf_mask_2000",   # <-- now exists
            tag=f"_{i:03d}",
            smooth_size=3,
        )

    # 4) Hydrology
    run_hydro_for_dem(
        dem_name=dem_for_hydro,
        out_dir=OUT,
        stream_threshold=STREAM_THRESHOLD,
        run_idx=i,
        smooth=False,
        perturbation_map=f"pert_mc_{i:03d}",
    )

    # 5) Optional exports (BEFORE cleanup)
    if EXPORT_DEBUG:
        if EXPORT_RUNS is not None:
            do_export = i in EXPORT_RUNS
        else:
            do_export = (i % EXPORT_EVERY == 0)

        if do_export:
            export_maps_for_run(i, OUT)

    # 6) Cleanup: remove anything created in this iteration (except the base maps)
    ras_after = list_maps("raster")
    vec_after = list_maps("vector")

    new_rasters = sorted((ras_after - ras_before) - KEEP_RASTERS)
    new_vectors = sorted((vec_after - vec_before) - KEEP_VECTORS)

    remove_maps("raster", new_rasters)
    remove_maps("vector", new_vectors)

# Optional postprocess:
# build_mc_products(OUT, basin_pattern=BASIN_PATTERN, chunk_rows=32)
```
