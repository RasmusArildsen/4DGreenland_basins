```python
import os
from collections import defaultdict

import basin_core as bc
from basin_core import setup_grass_env, start_grass_from_raster, QGIS_PREFIX_DEFAULT, ensure_grass_addon


def stitch_supra_to_sub_fullflow_fluxgate(
    dem: str,
    basins_sub: str,
    supra_mask: str | None,
    z_switch: float = 2000.0,
    gate_grow_radius: int = 5,       # pixels
    watershed_memory: int = 20000,
    use_segmode: bool = True,        # r.watershed -m (disk-based)
    use_sfd: bool = False,           # r.watershed -s (D8). Start False (MFD is robust)
    out_gate_subid: str = "gate_2000_subid",
    out_gate_clump: str = "gate_2000_clump",
    out_supra_raw: str = "supra_raw",
    out_supra_to_sub: str = "supra_to_subid",
    out_merged: str = "basins_merged_2000m",
    tmp: str = "tmp_stitch",
    fill_unassigned_with_sub: bool = True,
    cleanup: bool = False,
):
    gs = bc.gs
    safe = bc.safe

    gs.run_command("g.region", raster=dem)

    # 1) Clean sub basins (kill -9999/0 -> NULL)
    basins_clean = f"{tmp}_basins_sub_clean"
    safe(f"{basins_clean} = if({basins_sub} > 0, {basins_sub}, null())")

    # 2) Mask domain to sub basins so extent matches (and avoids outside junk)
    ice_mask = f"{tmp}_ice_mask"
    safe(f"{ice_mask} = if(!isnull({basins_clean}), 1, null())")
    gs.run_command("r.mask", raster=ice_mask, overwrite=True)

    # 3) Flow direction + accumulation on FULL DEM (stable)
    acc = f"{tmp}_acc_full"
    drn = f"{tmp}_drn_full"

    flags = ""
    if use_segmode:
        flags += "m"
    if use_sfd:
        flags += "s"

    gs.run_command(
        "r.watershed",
        elevation=dem,
        accumulation=acc,
        drainage=drn,
        memory=watershed_memory,
        flags=flags,
        overwrite=True,
    )

    # 4) Supra domain (>=2000 and optional supra_mask)
    supra_dom = f"{tmp}_supra_dom"
    if supra_mask:
        safe(
            f"{supra_dom} = if(!isnull({supra_mask}), "
            f"               if({dem} >= {z_switch}, 1, null()), "
            f"               null())"
        )
    else:
        safe(f"{supra_dom} = if({dem} >= {z_switch}, 1, null())")

    # 5) Build flux-gate candidate band: supra cells adjacent to sub-below band
    sub_below_id   = f"{tmp}_sub_below_id"
    sub_below_bin  = f"{tmp}_sub_below_bin"
    sub_below_grow = f"{tmp}_sub_below_grow"
    gate_bin       = f"{tmp}_gate_bin"

    safe(f"{sub_below_id}  = if({dem} < {z_switch}, {basins_clean}, null())")
    safe(f"{sub_below_bin} = if(!isnull({sub_below_id}), 1, null())")
    gs.run_command("r.grow", input=sub_below_bin, output=sub_below_grow,
                   radius=gate_grow_radius, overwrite=True)

    safe(f"{gate_bin} = if(!isnull({supra_dom}), if({sub_below_grow}, 1, null()), null())")

    print("Gate candidate cells (binary) univar:")
    print(gs.read_command("r.univar", map=gate_bin, flags="g"))

    # 6) Label gate cells with nearest sub-basin ID (within ~1 cell)
    sub_nn   = f"{tmp}_sub_nn"
    sub_dist = f"{tmp}_sub_dist"
    gs.run_command("r.grow.distance", input=sub_below_id, value=sub_nn, distance=sub_dist, overwrite=True)

    reg = gs.parse_command("g.region", flags="g")
    res = max(float(reg["nsres"]), float(reg["ewres"]))
    dmax = res * 1.6

    safe(f"{out_gate_subid} = if({gate_bin}, if({sub_dist} <= {dmax}, {sub_nn}, null()), null())")
    safe(f"{out_gate_subid} = if({out_gate_subid} > 0, {out_gate_subid}, null())")

    print("Gate outlets (subid) univar:")
    print(gs.read_command("r.univar", map=out_gate_subid, flags="g"))

    # 7) Make unique outlet IDs and delineate upstream areas
    gs.run_command("r.clump", input=out_gate_subid, output=out_gate_clump, overwrite=True)

    gs.run_command(
        "r.stream.basins",
        direction=drn,
        stream_rast=out_gate_clump,
        basins=out_supra_raw,
        overwrite=True,
    )

    # 8) Map supra_raw -> sub_id using only gate pixels (majority vote)
    supra_at_gate = f"{tmp}_supra_at_gate"
    sub_at_gate   = f"{tmp}_sub_at_gate"
    safe(f"{supra_at_gate} = if(!isnull({out_gate_clump}), {out_supra_raw}, null())")
    safe(f"{sub_at_gate}   = if(!isnull({out_gate_clump}), {out_gate_subid}, null())")

    stats = gs.read_command(
        "r.stats",
        flags="c",
        input=f"{supra_at_gate},{sub_at_gate}",
        separator=",",
        null_value="*",
    )

    counts: dict[int, dict[int, int]] = defaultdict(dict)
    for line in stats.strip().splitlines():
        a, b, c = line.split(",")
        if a in ("*", "") or b in ("*", ""):
            continue
        try:
            supra_id = int(a); sub_id = int(b); cnt = int(c)
        except ValueError:
            continue
        if supra_id <= 0 or sub_id <= 0 or cnt <= 0:
            continue
        counts.setdefault(supra_id, {})
        counts[supra_id][sub_id] = counts[supra_id].get(sub_id, 0) + cnt

    mapping = {sid: max(sc.items(), key=lambda kv: kv[1])[0] for sid, sc in counts.items()}
    print(f"Mapped supra regions: {len(mapping)}")

    rules = "\n".join([f"{k} = {v}" for k, v in mapping.items()]) + "\n* = 0\n"
    gs.write_command(
        "r.reclass",
        input=out_supra_raw,
        output=out_supra_to_sub,
        rules="-",
        stdin=rules,
        overwrite=True,
    )

    # Unmapped -> NULL, restrict to supra domain
    safe(f"{out_supra_to_sub} = if({out_supra_to_sub} > 0, {out_supra_to_sub}, null())")
    safe(f"{out_supra_to_sub} = if(!isnull({supra_dom}), {out_supra_to_sub}, null())")

    # Remove mask before final merge/export
    try:
        gs.run_command("r.mask", flags="r")
    except Exception:
        pass

    # 9) Merge: below 2000 unchanged sub; above 2000 stitched (fallback optional)
    if fill_unassigned_with_sub:
        safe(
            f"{out_merged} = if({dem} < {z_switch}, {basins_clean}, "
            f"               if(!isnull({out_supra_to_sub}), {out_supra_to_sub}, {basins_clean}))"
        )
    else:
        safe(f"{out_merged} = if({dem} < {z_switch}, {basins_clean}, {out_supra_to_sub})")

    if cleanup:
        tmp_maps = [
            basins_clean, ice_mask, acc, drn, supra_dom,
            sub_below_id, sub_below_bin, sub_below_grow, gate_bin,
            sub_nn, sub_dist, supra_at_gate, sub_at_gate,
        ]
        gs.run_command("g.remove", flags="f", type="rast", name=",".join(tmp_maps), quiet=True)

    print("✓ Stitch finished")
    print("  merged:", out_merged)
    print("  supra_to_sub:", out_supra_to_sub)


# =========================
# DRIVER (your paths)
# =========================

DEM_FULL = r"E:\Rasmus\DTU\Cryo\4DGreenland\DEM\arcticDEM_500m_ice_sheet.tif"
SUPRA_MASK_GPKG = r"E:\Rasmus\DTU\Cryo\4DGreenland\DEM\mask_2000m_largest.gpkg"
SUB_BASINS_TIF  = r"output\arcticdem_500_single_smooth_10\basins_hydro.tif"

setup_grass_env(QGIS_PREFIX_DEFAULT)
GISDBASE, LOCATION, MAPSET = start_grass_from_raster(DEM_FULL)

ensure_grass_addon("r.stream.basins")

bc.gs.run_command("r.in.gdal", input=DEM_FULL, output="dem", overwrite=True)
bc.gs.run_command("g.region", raster="dem")

bc.gs.run_command("r.in.gdal", input=SUB_BASINS_TIF, output="basins_sub", overwrite=True)

# supra mask
try:
    bc.gs.run_command("v.import", input=SUPRA_MASK_GPKG, output="supra_mask_vec", overwrite=True)
except Exception:
    bc.gs.run_command("v.in.ogr", input=SUPRA_MASK_GPKG, output="supra_mask_vec", overwrite=True)

bc.gs.run_command("v.to.rast", input="supra_mask_vec", output="supra_mask_rast",
                  use="val", value=1, overwrite=True)

# Run stitch (NOTE: use_sfd=False first; it’s more robust)
stitch_supra_to_sub_fullflow_fluxgate(
    dem="dem",
    basins_sub="basins_sub",
    supra_mask="supra_mask_rast",   # or None
    z_switch=2000.0,
    gate_grow_radius=5,
    watershed_memory=20000,
    use_segmode=True,
    use_sfd=False,
    out_merged="basins_merged_2000m",
    cleanup=False,
)

# Export
out_dir = r"output\arcticdem_500_single_smooth_10"
os.makedirs(out_dir, exist_ok=True)

bc.gs.run_command(
    "r.out.gdal",
    input="basins_merged_2000m",
    output=os.path.join(out_dir, "basins_merged_2000m.tif"),
    format="GTiff",
    createopt="COMPRESS=DEFLATE",
    overwrite=True,
)
print("✅ Exported:", os.path.join(out_dir, "basins_merged_2000m.tif"))
```
