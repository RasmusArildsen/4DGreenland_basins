```python
# run_single_hydro_100m.py

from pathlib import Path

import basin_core as bc
from basin_core import (
    setup_grass_env,
    start_grass_from_raster,
    prepare_dem_and_mask,
    ensure_grass_addon,
    run_hydro_for_dem,
    QGIS_PREFIX_DEFAULT,
)

# --------------------------------------------------------------------
# USER SETTINGS
# --------------------------------------------------------------------
DEM = r"E:\Rasmus\DTU\Cryo\4DGreenland\DEM\arcticDEM_500m_ice_sheet.tif"

# Use EXACTLY the same ice mask as in the MC runs (PROMICE polygon GPKG)
#ICE_MASK = r"E:\Rasmus\DTU\Cryo\4DGreenland\Ice_mask\02-PROMICE-2022-IceMask-polygon.gpkg"
ICE_MASK =r"E:\Rasmus\DTU\Cryo\4DGreenland\DEM\mask_2000m_largest.gpkg"   # your >=2000 largest component


OUT = r"output\arcticdem_500_2000m_v2"
STREAM_THRESHOLD = 1

# DEM hole filling / smoothing – keep it simple/basic
FILL_HOLES  = False        # same style as your MC script
FILL_METHOD = "bilinear"   # only used if FILL_HOLES=True

SMOOTH      = False        # no smoothing
SMOOTH_SIZE = 9            # ignored when SMOOTH=False

QGIS_PREFIX = QGIS_PREFIX_DEFAULT  # or your explicit QGIS path if you prefer

Path(OUT).mkdir(parents=True, exist_ok=True)

# --------------------------------------------------------------------
# INIT GRASS
# --------------------------------------------------------------------
setup_grass_env(QGIS_PREFIX)
gs = bc.gs  # just like in run_monte_carlo.py

GISDBASE, LOCATION, MAPSET = start_grass_from_raster(DEM)

ensure_grass_addon("r.stream.extract")
ensure_grass_addon("r.stream.basins")

# --------------------------------------------------------------------
# DEM + MASK  (vector PROMICE mask, same as MC runs)
# --------------------------------------------------------------------
# This:
#   - imports DEM as 'dem'
#   - optionally fills holes → 'dem_filled_100m' (if FILL_HOLES=True)
#   - imports PROMICE GPKG → ice_mask_vec → ice_mask_rast
#   - sets r.mask
# and returns (DEM_FOR_HYDRO, 'ice_mask_rast')
dem_map, ice_mask_rast = prepare_dem_and_mask(
    dem_path=DEM,
    ice_mask_path=ICE_MASK,
    fill_holes=FILL_HOLES,
    fill_method=FILL_METHOD,
    filled_name="dem_filled_100m",
)

# --------------------------------------------------------------------
# HYDRO RUN FOR THIS DEM (no perturbation, no smoothing)
# --------------------------------------------------------------------
run_hydro_for_dem(
    dem_name=dem_map,
    out_dir=OUT,
    stream_threshold=STREAM_THRESHOLD,
    run_idx=None,        # single run, not MC index
    smooth=SMOOTH,       # False → no smoothing
    smooth_size=SMOOTH_SIZE,
    perturbation_map=None,
)

print("✓ Single 100 m hydrology run finished")
print(f"  Output directory: {OUT}")
print(f"  DEM used:  {DEM}")
print(f"  ICE MASK:  {ICE_MASK}")
```
