```python
# batch_glacier_basin_ensemble_multi.py
from __future__ import annotations

from pathlib import Path
import numpy as np
import pandas as pd
import geopandas as gpd
import rasterio
from rasterio.windows import Window
from rasterio import features
from shapely.geometry import shape
from shapely.ops import unary_union
import matplotlib.pyplot as plt


# ---------------------------- CONFIG ----------------------------
MC_DIR = Path(r"/Volumes/Lenovo PS8/4DBasins/hybrid_500_mc2")
BASIN_GLOB = "basins_hydro_mc_*.tif"  # change if needed

MOST_LIKELY_NAME = "basins_most_likely.tif"
MERGED_NAME = "basins_most_likely_merged.tif"

GLACIERS_ROOT = Path(r"/Users/ralor/Rasmus/DTU/Cryo/4DGreenland")
GLACIERS = ["Helheim", "Ilulissat", "Midgard", "Peterman", "Ryder", "Serqqap", "Zach"]

FEATURE_INDEX = 0
CHUNK_ROWS = 512
HIST_BINS = 35

# Controls: how much overlap inside the glacier polygon is required to keep a basin label
MIN_POLY_COVER_PIX = 10     # drop tiny slivers (in pixels within polygon)
# ---------------------------------------------------------------


def _pixel_area_m2(ds: rasterio.io.DatasetReader) -> float:
    return float(abs(ds.transform.a * ds.transform.e))


def _count_pixels_for_labels_in_polygon(
    ds: rasterio.io.DatasetReader,
    geom,
    chunk_rows: int,
) -> dict[int, int]:
    """
    Returns {label: count_of_pixels_within_polygon} for all labels >0 within geom.
    Windowed to avoid reading full raster.
    """
    H, W = ds.height, ds.width
    nodata = ds.nodata
    counts: dict[int, int] = {}

    for r0 in range(0, H, chunk_rows):
        r1 = min(H, r0 + chunk_rows)
        win = Window(0, r0, W, r1 - r0)
        arr = ds.read(1, window=win)

        if nodata is not None:
            arr = np.where(arr == nodata, 0, arr)

        win_transform = ds.window_transform(win)

        # mask for polygon in this window
        mask = features.geometry_mask(
            [geom],
            out_shape=(r1 - r0, W),
            transform=win_transform,
            invert=True,  # True inside geometry
            all_touched=False,
        )
        if not np.any(mask):
            continue

        vals = arr[mask]
        vals = vals[vals > 0]
        if vals.size == 0:
            continue

        uniq, cnt = np.unique(vals, return_counts=True)
        for u, c in zip(uniq, cnt):
            u = int(u)
            counts[u] = counts.get(u, 0) + int(c)

    return counts


def _area_km2_for_label(ds: rasterio.io.DatasetReader, label: int, chunk_rows: int) -> float:
    if label <= 0:
        return float("nan")

    px_area = _pixel_area_m2(ds)
    H, W = ds.height, ds.width
    nodata = ds.nodata

    total = 0
    for r0 in range(0, H, chunk_rows):
        r1 = min(H, r0 + chunk_rows)
        win = Window(0, r0, W, r1 - r0)
        arr = ds.read(1, window=win)
        if nodata is not None:
            arr = np.where(arr == nodata, 0, arr)
        total += int(np.sum(arr == label))

    return float(total * px_area / 1e6)


def _best_run_label_for_ref_label_within_polygon(
    ref_label: int,
    ref_ds: rasterio.io.DatasetReader,
    run_path: Path,
    geom,
    chunk_rows: int,
) -> int:
    """
    Find the run label that best corresponds to ref_label, but only using pixels
    within the glacier polygon (geom) AND within ref_label area.
    Returns 0 if none.
    """
    best_label = 0
    best_count = 0

    H, W = ref_ds.height, ref_ds.width
    nodata_ref = ref_ds.nodata

    with rasterio.open(run_path) as run_ds:
        nodata_run = run_ds.nodata

        for r0 in range(0, H, chunk_rows):
            r1 = min(H, r0 + chunk_rows)
            win = Window(0, r0, W, r1 - r0)

            ref = ref_ds.read(1, window=win)
            run = run_ds.read(1, window=win)

            if nodata_ref is not None:
                ref = np.where(ref == nodata_ref, 0, ref)
            if nodata_run is not None:
                run = np.where(run == nodata_run, 0, run)

            # polygon mask in this window
            win_transform = ref_ds.window_transform(win)
            poly_mask = features.geometry_mask(
                [geom],
                out_shape=(r1 - r0, W),
                transform=win_transform,
                invert=True,
                all_touched=False,
            )
            if not np.any(poly_mask):
                continue

            mask = poly_mask & (ref == ref_label)
            if not np.any(mask):
                continue

            vals = run[mask]
            vals = vals[vals > 0]
            if vals.size == 0:
                continue

            uniq, cnt = np.unique(vals, return_counts=True)
            j = int(uniq[np.argmax(cnt)])
            c = int(cnt.max())

            if c > best_count:
                best_count = c
                best_label = j

    return int(best_label)


def _write_membership_probability_union(
    ref_grid: rasterio.io.DatasetReader,
    run_files: list[Path],
    run_label_sets: list[set[int]],
    out_tif: Path,
    chunk_rows: int,
) -> None:
    """
    For each run i, we have a set of run labels that correspond to the glacier's
    selected ref basins. We compute p(x) = fraction of runs where pixel is in ANY of those labels.
    """
    N = len(run_files)
    meta = ref_grid.meta.copy()
    meta.update(dtype="float32", nodata=0.0, count=1, compress="LZW")

    H, W = ref_grid.height, ref_grid.width
    out_tif.parent.mkdir(parents=True, exist_ok=True)

    with rasterio.open(out_tif, "w", **meta) as dst:
        for r0 in range(0, H, chunk_rows):
            r1 = min(H, r0 + chunk_rows)
            win = Window(0, r0, W, r1 - r0)

            counts = np.zeros((r1 - r0, W), dtype=np.uint16)

            for fp, labs in zip(run_files, run_label_sets):
                if not labs:
                    continue
                with rasterio.open(fp) as ds:
                    arr = ds.read(1, window=win)
                    if ds.nodata is not None:
                        arr = np.where(arr == ds.nodata, 0, arr)

                    # mark membership in any basin label for this run
                    m = np.zeros(arr.shape, dtype=bool)
                    for lab in labs:
                        if lab > 0:
                            m |= (arr == lab)
                    counts += m.astype(np.uint16)

            p = counts.astype(np.float32) / float(N)
            p[counts == 0] = 0.0
            dst.write(p.astype(np.float32), 1, window=win)


def _basin_outline_union_shp(
    tif_path: Path,
    labels: list[int],
    out_shp: Path,
    chunk_rows: int = 1024,
) -> None:
    """
    Polygonize union mask (raster in labels) -> dissolve -> save shapefile.
    """
    labels_set = set(int(l) for l in labels if int(l) > 0)
    if not labels_set:
        raise ValueError("No valid labels for outline.")

    out_shp.parent.mkdir(parents=True, exist_ok=True)

    with rasterio.open(tif_path) as ds:
        H, W = ds.height, ds.width
        nodata = ds.nodata
        geoms = []

        for r0 in range(0, H, chunk_rows):
            r1 = min(H, r0 + chunk_rows)
            win = Window(0, r0, W, r1 - r0)
            arr = ds.read(1, window=win)

            if nodata is not None:
                arr = np.where(arr == nodata, 0, arr)

            mask = np.isin(arr, list(labels_set))
            if not np.any(mask):
                continue

            win_transform = ds.window_transform(win)

            # polygonize boolean mask -> use uint8 image
            img = mask.astype(np.uint8)
            for geom, val in features.shapes(img, mask=mask, transform=win_transform):
                if int(val) == 1:
                    geoms.append(shape(geom))

        if not geoms:
            raise ValueError(f"No polygons found in {tif_path.name} for labels union.")

        dissolved = unary_union(geoms)
        gdf = gpd.GeoDataFrame({"n_labels": [len(labels_set)]}, geometry=[dissolved], crs=ds.crs)

    gdf.to_file(out_shp, driver="ESRI Shapefile")


def process_glacier(glacier_name: str) -> None:
    glacier_dir = GLACIERS_ROOT / glacier_name
    shp = glacier_dir / f"{glacier_name}.shp"
    if not shp.exists():
        print(f"[{glacier_name}] SKIP: missing shapefile: {shp}")
        return

    run_files = sorted(MC_DIR.glob(BASIN_GLOB))
    if not run_files:
        raise FileNotFoundError(f"No ensemble rasters found: {MC_DIR}/{BASIN_GLOB}")

    most_likely_tif = MC_DIR / MOST_LIKELY_NAME
    merged_tif = MC_DIR / MERGED_NAME
    if not most_likely_tif.exists():
        raise FileNotFoundError(f"Missing {MOST_LIKELY_NAME} in {MC_DIR}")
    if not merged_tif.exists():
        raise FileNotFoundError(f"Missing {MERGED_NAME} in {MC_DIR}")

    gdf = gpd.read_file(shp)
    if len(gdf) == 0:
        print(f"[{glacier_name}] SKIP: empty shapefile")
        return
    if gdf.crs is None:
        raise ValueError(f"Shapefile has no CRS: {shp}")

    geom_src = gdf.geometry.iloc[FEATURE_INDEX]

    # Reproject geometry to raster CRS
    with rasterio.open(most_likely_tif) as ds_ml:
        geom = gpd.GeoSeries([geom_src], crs=gdf.crs).to_crs(ds_ml.crs).iloc[0]

        # Find which basins overlap the glacier polygon (in most-likely raster)
        pix_counts = _count_pixels_for_labels_in_polygon(ds_ml, geom, chunk_rows=CHUNK_ROWS)
        # filter tiny slivers
        pix_counts = {lab: c for lab, c in pix_counts.items() if c >= MIN_POLY_COVER_PIX}
        if not pix_counts:
            print(f"[{glacier_name}] SKIP: polygon overlaps no basin labels in {MOST_LIKELY_NAME}")
            return

        ref_labels = sorted(pix_counts.keys())
        # For “Midgard covers two basins” this will become 2 labels.
        areas_ml = {lab: _area_km2_for_label(ds_ml, lab, chunk_rows=CHUNK_ROWS) for lab in ref_labels}
        area_ml_sum = float(np.sum(list(areas_ml.values())))

    # Do the same in merged raster: labels may have changed, so re-extract labels under polygon
    with rasterio.open(merged_tif) as ds_mg:
        pix_counts_mg = _count_pixels_for_labels_in_polygon(ds_mg, geom, chunk_rows=CHUNK_ROWS)
        pix_counts_mg = {lab: c for lab, c in pix_counts_mg.items() if c >= MIN_POLY_COVER_PIX}
        if not pix_counts_mg:
            print(f"[{glacier_name}] SKIP: polygon overlaps no basin labels in {MERGED_NAME}")
            return
        merged_labels = sorted(pix_counts_mg.keys())
        areas_mg = {lab: _area_km2_for_label(ds_mg, lab, chunk_rows=CHUNK_ROWS) for lab in merged_labels}
        area_mg_sum = float(np.sum(list(areas_mg.values())))

    print(f"\n[{glacier_name}] most_likely labels under polygon: {len(ref_labels)}  sum area={area_ml_sum:.1f} km²")
    print(f"[{glacier_name}] merged labels under polygon:      {len(merged_labels)}  sum area={area_mg_sum:.1f} km²")
    print(f"[{glacier_name}] ensemble runs: {len(run_files)}")

    # Ensemble areas: for each run, map each ref_label -> best run label (within polygon), then sum areas
    run_label_sets: list[set[int]] = []
    run_area_sums: list[float] = []

    with rasterio.open(most_likely_tif) as ref_ds:
        for i, fp in enumerate(run_files, start=1):
            run_labs = set()
            area_sum = 0.0

            # map each ref label independently
            for ref_lab in ref_labels:
                run_lab = _best_run_label_for_ref_label_within_polygon(
                    ref_label=ref_lab,
                    ref_ds=ref_ds,
                    run_path=fp,
                    geom=geom,
                    chunk_rows=CHUNK_ROWS,
                )
                if run_lab > 0:
                    run_labs.add(run_lab)

            # compute summed area in this run for the union of those run labels
            with rasterio.open(fp) as ds:
                for lab in run_labs:
                    area_sum += _area_km2_for_label(ds, lab, chunk_rows=CHUNK_ROWS)

            run_label_sets.append(run_labs)
            run_area_sums.append(area_sum)

            if i % 25 == 0 or i == len(run_files):
                print(f"[{glacier_name}] processed {i}/{len(run_files)}")

    glacier_dir.mkdir(parents=True, exist_ok=True)

    out_csv = glacier_dir / f"basin_area_ensemble_{glacier_name}.csv"
    out_png = glacier_dir / f"basin_area_hist_{glacier_name}.png"
    out_prob = glacier_dir / f"basin_membership_probability_{glacier_name}.tif"

    out_shp_ml = glacier_dir / f"basin_outline_most_likely_{glacier_name}.shp"
    out_shp_mg = glacier_dir / f"basin_outline_most_likely_merged_{glacier_name}.shp"

    # CSV
    df = pd.DataFrame(
        {
            "run_index": np.arange(1, len(run_files) + 1),
            "raster": [str(p) for p in run_files],
            "n_ref_labels_most_likely": len(ref_labels),
            "basin_area_sum_km2": run_area_sums,
            "run_labels_union": [",".join(map(str, sorted(s))) for s in run_label_sets],
        }
    )
    df.to_csv(out_csv, index=False)

    # Histogram + vertical markers for the SUM areas from most-likely and merged
    # Histogram + vertical markers for the SUM areas from most-likely and merged
    vals = df["basin_area_sum_km2"].dropna().values
    fig, ax = plt.subplots(figsize=(9, 5))
    ax.hist(vals, bins=HIST_BINS)
    
    ax.set_xlabel("Basin area (km²) (sum of basins intersecting glacier polygon)")
    ax.set_ylabel("Count (ensemble runs)")
    ax.set_title(f"{glacier_name}: basin area distribution across ensemble (multi-basin)")
    ax.grid(True, alpha=0.25)
    
    # --- colored reference lines + legend ---
    ax.axvline(
        area_ml_sum,
        linestyle="--",
        linewidth=2,
        color="orange",
        label=f"Most likely basins (sum={area_ml_sum:.1f} km²)",
    )
    ax.axvline(
        area_mg_sum,
        linestyle=":",
        linewidth=2,
        color="green",
        label=f"Most likely basins merged (sum={area_mg_sum:.1f} km²)",
    )
    
    ax.legend(loc="best")
    
    fig.tight_layout()
    fig.savefig(out_png, dpi=200)
    plt.close(fig)



def main():
    for g in GLACIERS:
        process_glacier(g)


if __name__ == "__main__":
    main()
```

```python
# batch_glacier_basin_ensemble_multi_mapped.py
from __future__ import annotations

from pathlib import Path
from collections import defaultdict

import numpy as np
import pandas as pd
import geopandas as gpd
import rasterio
from rasterio.windows import Window
from rasterio import features
import matplotlib.pyplot as plt


# ---------------------------- CONFIG ----------------------------
MC_DIR = Path(r"/Volumes/Lenovo PS8/4DBasins/hybrid_500_mc2")
BASIN_GLOB = "basins_hydro_mc_*.tif"

MOST_LIKELY_NAME = "basins_most_likely.tif"
MERGED_NAME = "basins_most_likely_merged.tif"

GLACIERS_ROOT = Path(r"/Users/ralor/Rasmus/DTU/Cryo/4DGreenland")
GLACIERS = ["Helheim", "Ilulissat", "Midgard", "Peterman", "Ryder", "Serqqap", "Zach"]

FEATURE_INDEX = 0
CHUNK_ROWS = 512
HIST_BINS = 35

MIN_POLY_COVER_PIX = 10  # drop tiny slivers under polygon
# ---------------------------------------------------------------


def _pixel_area_m2(ds: rasterio.io.DatasetReader) -> float:
    return float(abs(ds.transform.a * ds.transform.e))


def _count_pixels_for_labels_in_polygon(
    ds: rasterio.io.DatasetReader,
    geom,
    chunk_rows: int,
) -> dict[int, int]:
    """Returns {label: pixel_count_within_polygon} for all labels >0 within geom."""
    H, W = ds.height, ds.width
    nodata = ds.nodata
    counts: dict[int, int] = {}

    for r0 in range(0, H, chunk_rows):
        r1 = min(H, r0 + chunk_rows)
        win = Window(0, r0, W, r1 - r0)
        arr = ds.read(1, window=win)

        if nodata is not None:
            arr = np.where(arr == nodata, 0, arr)

        win_transform = ds.window_transform(win)

        mask = features.geometry_mask(
            [geom],
            out_shape=(r1 - r0, W),
            transform=win_transform,
            invert=True,
            all_touched=False,
        )
        if not np.any(mask):
            continue

        vals = arr[mask]
        vals = vals[vals > 0]
        if vals.size == 0:
            continue

        uniq, cnt = np.unique(vals, return_counts=True)
        for u, c in zip(uniq, cnt):
            u = int(u)
            counts[u] = counts.get(u, 0) + int(c)

    return counts


def _area_km2_for_label(ds: rasterio.io.DatasetReader, label: int, chunk_rows: int) -> float:
    if label <= 0:
        return float("nan")

    px_area = _pixel_area_m2(ds)
    H, W = ds.height, ds.width
    nodata = ds.nodata

    total = 0
    for r0 in range(0, H, chunk_rows):
        r1 = min(H, r0 + chunk_rows)
        win = Window(0, r0, W, r1 - r0)
        arr = ds.read(1, window=win)
        if nodata is not None:
            arr = np.where(arr == nodata, 0, arr)
        total += int(np.sum(arr == label))

    return float(total * px_area / 1e6)


def _max_label_in_raster(path: Path, chunk_rows: int) -> int:
    with rasterio.open(path) as ds:
        H, W = ds.height, ds.width
        mx = 0
        for r0 in range(0, H, chunk_rows):
            r1 = min(H, r0 + chunk_rows)
            win = Window(0, r0, W, r1 - r0)
            arr = ds.read(1, window=win)
            mx = max(mx, int(arr.max()))
        return int(mx)


def build_run_to_ref_mapping(
    ref_path: Path,
    run_path: Path,
    max_ref_label: int,
    chunk_rows: int,
) -> np.ndarray:
    """
    Build map_r such that map_r[run_label] = ref_label with max overlap.
    Same idea as build_mc_products mapping step.
    """
    # Determine max run label so we can size map_r
    max_run_label = _max_label_in_raster(run_path, chunk_rows=chunk_rows)
    map_r = np.zeros(max_run_label + 1, dtype=np.int32)
    map_r[0] = 0

    mapping_counts = defaultdict(int)

    with rasterio.open(ref_path) as ref_ds, rasterio.open(run_path) as run_ds:
        H, W = ref_ds.height, ref_ds.width
        nodata_ref = ref_ds.nodata if ref_ds.nodata is not None else 0
        nodata_run = run_ds.nodata if run_ds.nodata is not None else 0

        for r0 in range(0, H, chunk_rows):
            r1 = min(H, r0 + chunk_rows)
            win = Window(0, r0, W, r1 - r0)

            ref = ref_ds.read(1, window=win)
            run = run_ds.read(1, window=win)

            # match your basin_certainty domain logic: only where ref has basin
            domain = (ref != 0) & (ref != nodata_ref)
            mask = domain & (run != 0) & (run != nodata_run)
            if not np.any(mask):
                continue

            rvals = run[mask].astype(np.int64)
            kvals = ref[mask].astype(np.int64)

            idx_comb = rvals * (max_ref_label + 1) + kvals
            uniq, cnt = np.unique(idx_comb, return_counts=True)
            for u, c in zip(uniq, cnt):
                mapping_counts[int(u)] += int(c)

    if not mapping_counts:
        # no overlaps -> leave zeros
        return map_r

    best_k = np.zeros(max_run_label + 1, dtype=np.int32)
    best_n = np.zeros(max_run_label + 1, dtype=np.int64)

    for idx_comb, c in mapping_counts.items():
        j = idx_comb // (max_ref_label + 1)  # run label
        if j == 0:
            continue
        k = idx_comb % (max_ref_label + 1)   # ref label
        if c > best_n[j]:
            best_n[j] = c
            best_k[j] = k

    map_r[: len(best_k)] = best_k
    map_r[0] = 0
    return map_r


def write_membership_probability_refspace(
    ref_grid_path: Path,
    run_files: list[Path],
    maps: list[np.ndarray],
    ref_labels_set: set[int],
    out_tif: Path,
    chunk_rows: int,
) -> None:
    """
    Membership probability in REFERENCE LABEL SPACE:
      p(x) = fraction of runs where mapped_label(x) is in ref_labels_set
    """
    with rasterio.open(ref_grid_path) as ref_ds:
        meta = ref_ds.meta.copy()
        meta.update(dtype="float32", nodata=0.0, count=1, compress="LZW")

        H, W = ref_ds.height, ref_ds.width

        out_tif.parent.mkdir(parents=True, exist_ok=True)
        with rasterio.open(out_tif, "w", **meta) as dst:
            for r0 in range(0, H, chunk_rows):
                r1 = min(H, r0 + chunk_rows)
                win = Window(0, r0, W, r1 - r0)

                counts = np.zeros((r1 - r0, W), dtype=np.uint16)

                for fp, map_r in zip(run_files, maps):
                    with rasterio.open(fp) as ds:
                        arr = ds.read(1, window=win).astype(np.int32)
                        if ds.nodata is not None:
                            arr[arr == int(ds.nodata)] = 0
                        arr[arr < 0] = 0
                        arr[arr >= len(map_r)] = 0

                        labels_ref = map_r[arr]  # map to ref labels
                        counts += np.isin(labels_ref, list(ref_labels_set)).astype(np.uint16)

                p = counts.astype(np.float32) / float(len(run_files))
                p[counts == 0] = 0.0
                dst.write(p.astype(np.float32), 1, window=win)


def area_sum_per_run_refspace(
    run_path: Path,
    map_r: np.ndarray,
    ref_labels_set: set[int],
    chunk_rows: int,
    px_area_m2: float,
) -> float:
    """Sum area in a run for pixels whose mapped label is in ref_labels_set."""
    with rasterio.open(run_path) as ds:
        H, W = ds.height, ds.width
        total = 0

        for r0 in range(0, H, chunk_rows):
            r1 = min(H, r0 + chunk_rows)
            win = Window(0, r0, W, r1 - r0)

            arr = ds.read(1, window=win).astype(np.int32)
            if ds.nodata is not None:
                arr[arr == int(ds.nodata)] = 0
            arr[arr < 0] = 0
            arr[arr >= len(map_r)] = 0

            labels_ref = map_r[arr]
            total += int(np.sum(np.isin(labels_ref, list(ref_labels_set))))

    return float(total * px_area_m2 / 1e6)


def process_glacier(glacier_name: str) -> None:
    glacier_dir = GLACIERS_ROOT / glacier_name
    shp = glacier_dir / f"{glacier_name}.shp"
    if not shp.exists():
        print(f"[{glacier_name}] SKIP: missing shapefile: {shp}")
        return

    run_files = sorted(MC_DIR.glob(BASIN_GLOB))
    if not run_files:
        raise FileNotFoundError(f"No ensemble rasters found: {MC_DIR}/{BASIN_GLOB}")

    most_likely_tif = MC_DIR / MOST_LIKELY_NAME
    merged_tif = MC_DIR / MERGED_NAME
    if not most_likely_tif.exists():
        raise FileNotFoundError(f"Missing {MOST_LIKELY_NAME} in {MC_DIR}")
    if not merged_tif.exists():
        raise FileNotFoundError(f"Missing {MERGED_NAME} in {MC_DIR}")

    gdf = gpd.read_file(shp)
    if len(gdf) == 0:
        print(f"[{glacier_name}] SKIP: empty shapefile")
        return
    if gdf.crs is None:
        raise ValueError(f"Shapefile has no CRS: {shp}")

    geom_src = gdf.geometry.iloc[FEATURE_INDEX]

    # Reference labels under polygon (from most-likely)
    with rasterio.open(most_likely_tif) as ds_ml:
        geom = gpd.GeoSeries([geom_src], crs=gdf.crs).to_crs(ds_ml.crs).iloc[0]

        pix_counts = _count_pixels_for_labels_in_polygon(ds_ml, geom, chunk_rows=CHUNK_ROWS)
        pix_counts = {lab: c for lab, c in pix_counts.items() if c >= MIN_POLY_COVER_PIX}
        if not pix_counts:
            print(f"[{glacier_name}] SKIP: polygon overlaps no basin labels in {MOST_LIKELY_NAME}")
            return

        ref_labels = sorted(pix_counts.keys())
        ref_labels_set = set(ref_labels)

        areas_ml = {lab: _area_km2_for_label(ds_ml, lab, chunk_rows=CHUNK_ROWS) for lab in ref_labels}
        area_ml_sum = float(np.sum(list(areas_ml.values())))

        # for later: pixel area for converting mapped pixel counts to km2
        px_area_m2 = _pixel_area_m2(ds_ml)

        # Also need max ref label for mapping arrays
        max_ref_label = 0
        H, W = ds_ml.height, ds_ml.width
        for r0 in range(0, H, CHUNK_ROWS):
            r1 = min(H, r0 + CHUNK_ROWS)
            win = Window(0, r0, W, r1 - r0)
            arr = ds_ml.read(1, window=win)
            max_ref_label = max(max_ref_label, int(arr.max()))

    # Merged sum area line (based on merged labels under polygon)
    with rasterio.open(merged_tif) as ds_mg:
        pix_counts_mg = _count_pixels_for_labels_in_polygon(ds_mg, geom, chunk_rows=CHUNK_ROWS)
        pix_counts_mg = {lab: c for lab, c in pix_counts_mg.items() if c >= MIN_POLY_COVER_PIX}
        if not pix_counts_mg:
            print(f"[{glacier_name}] SKIP: polygon overlaps no basin labels in {MERGED_NAME}")
            return
        merged_labels = sorted(pix_counts_mg.keys())
        areas_mg = {lab: _area_km2_for_label(ds_mg, lab, chunk_rows=CHUNK_ROWS) for lab in merged_labels}
        area_mg_sum = float(np.sum(list(areas_mg.values())))

    print(f"\n[{glacier_name}] most_likely ref-labels under polygon: {len(ref_labels)}  sum area={area_ml_sum:.1f} km²")
    print(f"[{glacier_name}] merged labels under polygon:          {len(merged_labels)}  sum area={area_mg_sum:.1f} km²")
    print(f"[{glacier_name}] ensemble runs: {len(run_files)}")

    # ------------------- BUILD mappings like basin_certainty -------------------
    # maps[i][run_label] -> ref_label
    maps: list[np.ndarray] = []
    for i, fp in enumerate(run_files, start=1):
        map_r = build_run_to_ref_mapping(
            ref_path=most_likely_tif,
            run_path=fp,
            max_ref_label=max_ref_label,
            chunk_rows=CHUNK_ROWS,
        )
        maps.append(map_r)
        if i % 25 == 0 or i == len(run_files):
            print(f"[{glacier_name}] built mapping {i}/{len(run_files)}")

    # ------------------- Ensemble area in ref-label space -------------------
    run_area_sums: list[float] = []
    for i, (fp, map_r) in enumerate(zip(run_files, maps), start=1):
        area_sum = area_sum_per_run_refspace(
            run_path=fp,
            map_r=map_r,
            ref_labels_set=ref_labels_set,
            chunk_rows=CHUNK_ROWS,
            px_area_m2=px_area_m2,
        )
        run_area_sums.append(area_sum)
        if i % 25 == 0 or i == len(run_files):
            print(f"[{glacier_name}] processed (area) {i}/{len(run_files)}")

    # Outputs
    glacier_dir.mkdir(parents=True, exist_ok=True)
    out_csv = glacier_dir / f"basin_area_ensemble_{glacier_name}.csv"
    out_png = glacier_dir / f"basin_area_hist_{glacier_name}.png"
    out_prob = glacier_dir / f"basin_membership_probability_{glacier_name}.tif"

    df = pd.DataFrame(
        {
            "run_index": np.arange(1, len(run_files) + 1),
            "raster": [str(p) for p in run_files],
            "n_ref_labels_most_likely": len(ref_labels),
            "ref_labels_union": ",".join(map(str, ref_labels)),
            "basin_area_sum_km2": run_area_sums,
        }
    )
    df.to_csv(out_csv, index=False)

    # Histogram with orange/green lines + legend
    vals = df["basin_area_sum_km2"].dropna().values
    fig, ax = plt.subplots(figsize=(9, 5))
    ax.hist(vals, bins=HIST_BINS)

    ax.set_xlabel("Basin area (km²) (sum of ref basins intersecting glacier polygon)")
    ax.set_ylabel("Count (ensemble runs)")
    ax.set_title(f"{glacier_name}: basin area distribution across ensemble (mapped to ref labels)")
    ax.grid(True, alpha=0.25)

    ax.axvline(
        area_ml_sum,
        linestyle="--",
        linewidth=2,
        color="orange",
        label=f"Most likely basins (sum={area_ml_sum:.1f} km²)",
    )
    ax.axvline(
        area_mg_sum,
        linestyle=":",
        linewidth=2,
        color="green",
        label=f"Most likely basins merged (sum={area_mg_sum:.1f} km²)",
    )
    ax.legend(loc="best")

    fig.tight_layout()
    fig.savefig(out_png, dpi=200)
    plt.close(fig)

    # Membership probability raster in ref-label space
    write_membership_probability_refspace(
        ref_grid_path=most_likely_tif,   # grid/extent
        run_files=run_files,
        maps=maps,
        ref_labels_set=ref_labels_set,
        out_tif=out_prob,
        chunk_rows=CHUNK_ROWS,
    )

    print(f"[{glacier_name}] wrote: {out_csv}")
    print(f"[{glacier_name}] wrote: {out_png}")
    print(f"[{glacier_name}] wrote: {out_prob}")


def main():
    for g in GLACIERS:
        process_glacier(g)


if __name__ == "__main__":
    main()
```
